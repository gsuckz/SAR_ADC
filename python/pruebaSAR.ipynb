```python
def sar_adc_differential(vin_pos, vin_neg, n_bits, vref):
    """
    Simula un SAR ADC diferencial de n bits y devuelve el resultado en binario.

    Parámetros:
    vin_pos -- tensión en la entrada positiva (V)
    vin_neg -- tensión en la entrada negativa (V)
    n_bits -- resolución del ADC (bits)
    vref -- voltaje de referencia máxima (positivo)

    Retorna:
    String de n_bits representando el código binario de salida.
    """
    # Calcular la tensión diferencial
    v_diff = vin_pos - vin_neg

    # Saturar la entrada al rango [-Vref, +Vref]
    v_diff = max(-vref, min(v_diff, vref))

    # Convertir a código digital en rango [0, 2^n_bits - 1]
    code_max = (1 << n_bits) - 1  # 2^n_bits - 1
    digital_code = int(round((v_diff + vref) * code_max / (2 * vref)))

    # Limitar a rango válido
    digital_code = max(0, min(code_max, digital_code))

    # Convertir a binario con ceros a la izquierda
    binary_code = format(digital_code, f'0{n_bits}b')

    return binary_code
```

```python
def detectar_errores_sar(output_sim, expected_output):
    """
    Compara dos cadenas binarias y detecta los bits con error.

    Parámetros:
    output_sim -- string binario de la simulación (ej. '10101100')
    expected_output -- string binario esperado (ej. '10111100')

    Retorna:
    - Lista con los índices de los bits con error (0 es el MSB)
    - String con flechas (^) marcando la posición de los errores
    """
    if len(output_sim) != len(expected_output):
        raise ValueError("Las cadenas binarias deben tener la misma longitud.")

    errores = []
    marcas = []

    for i, (bit_sim, bit_exp) in enumerate(zip(output_sim, expected_output)):
        if bit_sim != bit_exp:
            errores.append(i)
            marcas.append('^')
        else:
            marcas.append(' ')

    marcas_str = ''.join(marcas)

    return errores, marcas_str
```

```python
sim = "100000111111"
esperado = sar_adc_differential(1.8,1.7,12,3.3)

errores, marcas = detectar_errores_sar(sim, esperado)

print("Simulación: ", sim)
print("Esperado:   ", esperado)
print("Errores:    ", marcas)
print("Índices con error:", errores)
```
